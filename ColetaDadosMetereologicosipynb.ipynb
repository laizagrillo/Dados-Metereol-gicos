```python
# https://docs.google.com/spreadsheets/d/1vW4hmVmkvqXEQFC14FMRw_rmg8EUFvf4e-IxR0ZJ0oM/edit?usp=sharing

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1vW4hmVmkvqXEQFC14FMRw_rmg8EUFvf4e-IxR0ZJ0oM/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print("=-=-=-=-=-=-=-=- DADOS COLETADOS =-=-=-=-=-=-=-=-")
dados_planilha = pd.read_csv(endpoint_planilha)
display(dados_planilha)
```

```python
# iterar cada resposta do formulário
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros = {
      'latitude':lat,
      'longitude':lon,
      'current':'temperature_2m'
  }

  resposta = requests.get(endpoint_api, params=parametros).json()

  temperatura = resposta['current']['temperature_2m']
  print(f"{nome_local}: {temperatura}°C")
```
